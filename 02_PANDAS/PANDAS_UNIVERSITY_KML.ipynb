```python
import pandas as pd

FILE_NAME = '../data/공공대학교, 사립대학교 차이점.csv'


df = pd.read_csv(FILE_NAME)

## 데이터 확인
df.head(3)
```

[2] 전처리 <hr>

### [2-1] 송희 Part

```python
# Tuition per Semester(한학기 등록금) 열 전처리
# 1) 값 대체: replace
# 2) 타입 변환: object --> float64
df['Tuition per Semester'].value_counts()
```

```python
tuition_fee = {'Below 50k': 25,
               '50k - 100k': 75,
               '100k - 150k': 125,
               '150k - 200k': 175,
               'Above 200k': 225}

df['Tuition per Semester'] = df['Tuition per Semester'].replace(tuition_fee)
df['Tuition per Semester'] = df['Tuition per Semester'].astype('float64')

df['Tuition per Semester']
```

```python
# Difficulty Paying Tuition(등록금 납부 어려움 정도) 열 전처리
# 1) 값 대체: replace
# 2) 타입 변환: object --> float64
df['Difficulty Paying Tuition'].unique()
```

```python
# 쉬움 ~ 어려움 = 1 ~ 5
tuition_difficulty = {"Very easy (I do not face any financial difficulty in paying my fees)": 1,
                  "Easy (Manage to pay without much difficulty)": 2,
                  "Moderate (Occasionally experience financial strain, but manageable)": 3,
                  "Difficult (Often face financial stress, but manage with some support)": 4,
                  "Very difficult (Struggling to cover fees, require loans or assistance)": 5}


df['Difficulty Paying Tuition'] = df['Difficulty Paying Tuition'].replace(tuition_difficulty)
df['Difficulty Paying Tuition'] = df['Difficulty Paying Tuition'].astype('float64')
df['Difficulty Paying Tuition']
```

```python
# Difficulty Paying Tuition(등록금 납부 어려움 정도) 열 전처리
# 1) 값 대체: replace
# 2) 타입 변환: object --> float64
df['\tScholarship/Financial Aid'].unique()
```

```python
df['\tScholarship/Financial Aid'] = df['\tScholarship/Financial Aid'].replace('Not sure',0)

df['\tScholarship/Financial Aid'].unique()
```

```python
scholarship = {'Yes, I am currently receiving financial aid or a scholarship.': 1,
               'Yes, but I do not qualify for financial aid or a scholarship.': 0,
               'No, my university does not offer financial aid or scholarships.': 0}

df['\tScholarship/Financial Aid'] = df['\tScholarship/Financial Aid'].replace(scholarship)
```

### [2-2] 동건 Part

```python
# Type of Previous School의 행을 수정
# 'Public School (Government-funded)'를 'Public'으로,
# 'Private School'을 'Private'으로 변경.

## public -> 1
## private -> 3

df['Type of Previous School'] = df['Type of Previous School'].replace({
        'Public School (Government-funded)': 'Public',
        'Private School': 'Private'
    })

df['Type of Previous School'] = df['Type of Previous School'].replace({
        'Public': 1,
        'Private':2,
        'Homeschooling':pd.NA
    })
```

```python
# Current School Type의 행을 수정.
# 'Public (Government-funded)'를 'Public'으로,
# 'Not currently enrolled'을 'Not enrolled'으로 변경

df['Current School Type'] = df['Current School Type'].replace({
        'Public (Government-funded)': 'Public',
        'Not currently enrolled': 'Not enrolled'
    })

# 등록X -> 결측치
# public -> 1
# private -> 2

df['Current School Type'] = df['Current School Type'].replace({
        'Public': 1,
        'Private': 2,
        'Not enrolled':pd.NA
    })
```

### [2-3] 재혁 Part <hs>

```python
## GPA 구간을 수치형으로 변화
gpa_map = {"1.0 - 2.0" : 1,
           "2.1 - 2.5" : 1,
		   "2.6 - 3.0" : 2,
		   "3.1 - 3.5" : 3,
		   "3.6 - 4.0" : 3}

df['Average GPA (CGPA)'] = df['Average GPA (CGPA)'].map(gpa_map)
df['Average GPA (CGPA)'] = df['Average GPA (CGPA)'].astype('float64')
df['Average GPA (CGPA)']
```

```python
## 스트레스 수준을 수치형으로 변화
stress_map = {"No stress at all" : 1,
              "Low stress" : 2,
			  "Moderate stress" : 3,
			  "High stress" : 4,
			  "Very high stress" : 5}
df['Academic Stress Level'] = df['Academic Stress Level'].map(stress_map)
df['Academic Stress Level'] = df['Academic Stress Level'].astype('float64')
df['Academic Stress Level']
```

```python
## 교수지원접근성을 수치형으로 변화
sup_app_map = {"Not supportive or approachable at all" : 1,
		   "Not very supportive or approachable" : 2,
		   "Neutral (Supportive at times, less approachable)" : 3,
		   "Supportive but not very approachable" : 4,
		   "Very supportive and approachable" : 5}
df['Professor Support/Accessibility'] = df['Professor Support/Accessibility'].map(sup_app_map)
df['Professor Support/Accessibility'] = df['Professor Support/Accessibility'].astype('float64')
df['Professor Support/Accessibility']
```

### [2-4] 동규 Part

```python
## [3-1] Extracurricular Participation Frequency(교외활동참여빈도) 자료형 및 수치 변환
## ------------------------------------------
# 'Never': 0
# 'Rarely': 1
# 'Occasionally': 2
# 'Yes, occasionally': 2 (Occasionally와 동일)
# 'Frequently': 3
# 'Very frequently': 4
## ------------------------------------------
# 자료형 object -> category 변환
df['Extracurricular Participation Frequency']=df['Extracurricular Participation Frequency'].astype('category')

# 수치 변환
for data in df['Extracurricular Participation Frequency']:
    if data == 'Never':
        df['Extracurricular Participation Frequency']=df['Extracurricular Participation Frequency'].replace('Never', 0)
    elif data == 'Rarely':
        df['Extracurricular Participation Frequency']=df['Extracurricular Participation Frequency'].replace('Rarely', 1)
    elif data == 'Occasionally':
        df['Extracurricular Participation Frequency']=df['Extracurricular Participation Frequency'].replace('Occasionally', 2)
    elif data == 'Yes, occasionally':
        df['Extracurricular Participation Frequency']=df['Extracurricular Participation Frequency'].replace('Yes, occasionally', 2)
    elif data == 'Frequently':
        df['Extracurricular Participation Frequency']=df['Extracurricular Participation Frequency'].replace('Frequently', 3)
    elif data == 'Very frequently':
        df['Extracurricular Participation Frequency']=df['Extracurricular Participation Frequency'].replace('Very frequently', 4)
print(f'dataDF["Extracurricular Participation Frequency"] dtype : {df["Extracurricular Participation Frequency"].dtype}')
df.head()
```

```python
## [3-2] Post-Graduation Plan(졸업후 계획) 수치 변환
# 1. 데이터 확인
print('unique : ', df['Post-Graduation Plan'].unique())
print('nunique : ', df['Post-Graduation Plan'].nunique(), '개')
print('dtype : ', df['Post-Graduation Plan'].dtype)

## ------------------------------------------
# 1. 취업 (Seeking employment) -> 취업 또는 실무 중심
# - "Seeking employment"
# - "Web Developer"
# - "First job for experience then business"
# 2. 창업 (Entrepreneurship) -> 직접 창업 의도 or 진행 중
# - "Starting a business"
# - "Already running an agency: motioncutproductions.com"
# - "I'll start my own business"
# 3. 유학/해외 진출 (Abroad) -> 졸업 후 해외 진출
# - "Abroad"
# - "Out of Country"
# 4. 학업 지속 (Further Education) -> 대학원 진학 등
# - "Planning to pursue further education"
# - "Further education and starting a business"
# 5. 무계획/모름 (Undecided/None)
# - "Nothing"  -> 뚜렷한 계획 없음
# - "Nothing Special right now"
# - "IDK"
# - "Not sure..."
# 6. 부정적 응답/조롱 (Dismissive) -> 조롱 or 비꼼성 응답
# - "Nothing, just wanted to ruin my life. so i enrolled here"
## ------------------------------------------
```

```python
## 2. 자료형 및 데이터 -> 범주형 변환
# 자료형 object -> category 변환
df['Post-Graduation Plan']=df['Post-Graduation Plan'].astype('category')

# 범주형 변환
# 새로운 컬럼(categorical_Post-Graduation Plan) 생성
df['categorical_PGP'] = pd.NA

df.loc[df['Post-Graduation Plan'].isin([
    "Seeking employment", "Web Developer", "First job for experience then business "
]), 'categorical_PGP'] = 'Seeking employment'

df.loc[df['Post-Graduation Plan'].isin([
    "Starting a business", "Already running an agency: motioncutproductions.com",
    "Not sure, if I'll wait for my degree. I'll start my own business. "
]), 'categorical_PGP'] = 'Entrepreneurship'

df.loc[df['Post-Graduation Plan'].isin([
    "Abroad", "Out of Country"
]), 'categorical_PGP'] = 'Abroad'

df.loc[df['Post-Graduation Plan'].isin([
    "Planning to pursue further education (e.g., graduate or professional studies)",
    "Further education and starting a business "
]), 'categorical_PGP'] = 'Further Education'

df.loc[df['Post-Graduation Plan'].isin([
    "Nothing ", "Nothing Special right now", "IDK"
]), 'categorical_PGP'] = 'Undecided/None'

df.loc[df['Post-Graduation Plan'] == "Nothing,just wanted to ruin my life.so i enrolled here", 'categorical_PGP'] = 'Dismissive'
print(f'dataDF["Post-Graduation Plan"] dtype : {df["Post-Graduation Plan"].dtype}')
```
